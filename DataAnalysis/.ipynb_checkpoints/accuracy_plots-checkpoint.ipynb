```python
from random import *
import os
import glob
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import pandas as pd
import numpy.ma as ma
import pickle
import seaborn as sns
import sys
import pathlib
import itertools
import re
```

##### Individual learning curves with moving average of 5

```python
def ma(interval, window_size, method):
    window = np.ones(int(window_size))/float(window_size)
    return np.convolve(interval, window, method)
```

```python
def plotAccuracy(data, info, ww, method):
    data_block1 = data[data.runNumber == 1]
    data_block2 = data[data.runNumber == 2]
    data_block3 = data[data.runNumber == 3].reset_index(drop=True)
    data_block4 = data[data.runNumber == 4].reset_index(drop=True)

    participantID = info.participantID[0]

    fig, ax = plt.subplots(1, 2, sharex='col', sharey='row', figsize=(10,4))

    if info.beeps[0] == 1:
        plt.suptitle('ParticipantID: {}, Beeps'.format(participantID))
    else:
        plt.suptitle('ParticipantID: {}, No beeps'.format(participantID))

    if ~np.isnan(data_block1.audio[0]):
        audio_df = pd.concat((data_block1[['accurate', 'reward']][0:20], data_block1[['accurate', 'reward']][40:60], data_block2[['accurate', 'reward']][20:40]), ignore_index=True)
        audio_acc = pd.DataFrame(ma(audio_df.accurate, ww, method)).fillna(method='ffill')

        tactile_df = pd.concat((data_block1[['accurate', 'reward']][20:40], data_block2[['accurate', 'reward']][0:20], data_block2[['accurate', 'reward']][40:60]), ignore_index=True)
        tactile_acc = pd.DataFrame(ma(tactile_df.accurate, ww, method)).fillna(method='ffill')

        ax[0].plot(audio_acc[0:int(audio_acc.shape[0]/3)], 'g')
        ax[0].plot(audio_acc[int(audio_acc.shape[0]/3):int(audio_acc.shape[0]/3*2)], 'g')
        ax[0].plot(audio_acc[int(audio_acc.shape[0]/3*2):int(audio_acc.shape[0])], 'r')
        ax[0].set_title("Audio-Visual accuracy (moving avg: {})".format(ww))
        ax[0].set_ylabel('accuracy')
        ax[0].set_ylim(0, 1.1)

        ax[1].plot(tactile_acc[0:int(tactile_acc.shape[0]/3)], 'g')
        ax[1].plot(tactile_acc[int(tactile_acc.shape[0]/3):int(tactile_acc.shape[0]/3*2)], 'r')
        ax[1].plot(tactile_acc[int(tactile_acc.shape[0]/3*2):int(tactile_acc.shape[0])], 'r')
        ax[1].set_title("Tactile-Visual accuracy (moving avg: {})".format(ww))
        ax[1].set_ylabel('accuracy')
        ax[1].set_ylim(0, 1.1)

        if method == 'same':
            tactile_lines = tactile_df[(tactile_df.accurate != tactile_df.reward) & (~np.isnan(tactile_df.reward))].index.tolist()
            audio_lines = audio_df[(audio_df.accurate != audio_df.reward) & (~np.isnan(audio_df.reward))].index.tolist()

            for xc in audio_lines:
                ax[0].axvline(x=xc, color='k', linestyle='--')
            for xc in tactile_lines:
                ax[1].axvline(x=xc, color='k', linestyle='--')



    else:
        audio_df = pd.concat((data_block1[['accurate', 'reward']][20:40], data_block2[['accurate', 'reward']][0:20], data_block2[['accurate', 'reward']][40:60]), ignore_index=True)
        audio_acc = pd.DataFrame(ma(audio_df.accurate, ww, method)).fillna(method='ffill')

        tactile_df = pd.concat((data_block1[['accurate', 'reward']][0:20], data_block1[['accurate', 'reward']][40:60], data_block2[['accurate', 'reward']][20:40]), ignore_index=True)
        tactile_acc = pd.DataFrame(ma(tactile_df.accurate, ww, method)).fillna(method='ffill')

        ax[0].plot(tactile_acc[0:int(tactile_acc.shape[0]/3)], 'g')
        ax[0].plot(tactile_acc[int(tactile_acc.shape[0]/3):int(tactile_acc.shape[0]/3*2)], 'g')
        ax[0].plot(tactile_acc[int(tactile_acc.shape[0]/3*2):int(tactile_acc.shape[0])], 'r')
        ax[0].set_title("Tactile-Visual accuracy (moving avg: {})".format(ww))
        ax[0].set_ylabel('accuracy')
        ax[0].set_ylim(0, 1.1)

        ax[1].plot(audio_acc[0:int(audio_acc.shape[0]/3)], 'g')
        ax[1].plot(audio_acc[int(audio_acc.shape[0]/3):int(audio_acc.shape[0]/3*2)], 'r')
        ax[1].plot(audio_acc[int(audio_acc.shape[0]/3*2):int(audio_acc.shape[0])], 'r')
        ax[1].set_title("Audio-Visual accuracy (moving avg: {})".format(ww))
        ax[1].set_ylabel('accuracy')
        ax[1].set_ylim(0, 1.1)

        if method == 'same':
            tactile_lines = tactile_df[(tactile_df.accurate != tactile_df.reward) & (~np.isnan(tactile_df.reward))].index.tolist()
            audio_lines = audio_df[(audio_df.accurate != audio_df.reward) & (~np.isnan(audio_df.reward))].index.tolist()

            for xc in audio_lines:
                ax[1].axvline(x=xc, color='k', linestyle='--', linewidth=1)
            for xc in tactile_lines:
                ax[0].axvline(x=xc, color='k', linestyle='--', linewidth=1)

    fig2, ax2 = plt.subplots(1, 2, sharex='col', sharey='row',figsize=(10,4))

    if ~np.isnan(data_block3.audio[59]):
        ax2[0].plot(pd.DataFrame(ma(data_block3.accurate,ww, method)).fillna(method='ffill'))
        ax2[0].set_title("Audio-Visual accuracy (same block, moving avg: {})".format(ww))
        ax2[0].set_ylabel('accuracy')
        ax2[0].set_ylim(0, 1.1)

        ax2[1].plot(pd.DataFrame(ma(data_block4.accurate,ww, method)).fillna(method='ffill'))
        ax2[1].set_title("Tactile-Visual accuracy (same block, moving avg: {})".format(ww))
        ax2[1].set_ylabel('accuracy')
        ax2[1].set_ylim(0, 1.1)

        if method == 'same':
            audio_lines = data_block3[(data_block3.accurate != data_block3.reward) & (~np.isnan(data_block3.reward))].index.tolist()
            tactile_lines = data_block4[(data_block4.accurate != data_block4.reward) & (~np.isnan(data_block4.reward))].index.tolist()

            for xc in audio_lines:
                ax2[0].axvline(x=xc, color='k', linestyle='--', linewidth=1)
            for xc in tactile_lines:
                ax2[1].axvline(x=xc, color='k', linestyle='--', linewidth=1)

    else:
        ax2[0].plot(pd.DataFrame(ma(data_block3.accurate,ww, method)).fillna(method='ffill'))
        ax2[0].set_title("Tactile-Visual accuracy (same block, moving avg: {})".format(ww))
        ax2[0].set_ylabel('accuracy')
        ax2[0].set_ylim(0, 1.1)

        ax2[1].plot(pd.DataFrame(ma(data_block4.accurate,ww, method)).fillna(method='ffill'))
        ax2[1].set_title("Audio-Visual accuracy (same block, moving avg: {})".format(ww))
        ax2[1].set_ylabel('accuracy')
        ax2[1].set_ylim(0, 1.1)

        if method == 'same':
            tactile_lines = data_block3[(data_block3.accurate != data_block3.reward) & (~np.isnan(data_block3.reward))].index.tolist()
            audio_lines = data_block4[(data_block4.accurate != data_block4.reward) & (~np.isnan(data_block4.reward))].index.tolist()

            for xc in audio_lines:
                ax2[1].axvline(x=xc, color='k', linestyle='--', linewidth=1)
            for xc in tactile_lines:
                ax2[0].axvline(x=xc, color='k', linestyle='--', linewidth=1)

    plt.show()

```

##### Mean RT group analysis

```python
def plotGroupRT(folder='Data/Pilot/beepStimuli', exclude=False, group=False):

    dir_file = pathlib.Path().absolute().parent.parent.resolve()
    wanted_dir = os.path.join(dir_file, folder)

    # Get all expInfo and savedVals files
    files = [os.path.join(root, name) for root, dirs, files in os.walk(wanted_dir) for name in files if name.endswith('savedValues.csv')]

    # Loop through files and store RT vals, audio, tactile with probability per run and trial type
    run1and2 = list()
    run3and4 = list()

    for file in files:
        data = pd.read_csv(file)
        oneAndtwo = data[(data.runNumber == 1) | (data.runNumber == 2)][['responseTime', 'audio', 'tactile', 'combinationConditionalProbability']]
        threeAndfour = data[(data.runNumber == 3) | (data.runNumber == 4)][['responseTime', 'audio', 'tactile', 'combinationConditionalProbability']]

        if exclude:
            one_acc = np.array(data[data.runNumber == 1][['totalAccuracy']])[-1]
            two_acc = np.array(data[data.runNumber == 2][['totalAccuracy']])[-1]
            three_acc = np.array(data[data.runNumber == 3][['totalAccuracy']])[-1]
            four_acc = np.array(data[data.runNumber == 4][['totalAccuracy']])[-1]
            avg_acc = np.mean((one_acc, two_acc,three_acc, four_acc))
            if avg_acc >= 0.6:
                run1and2.append(oneAndtwo)
                run3and4.append(threeAndfour)
        else:
            run1and2.append(oneAndtwo)
            run3and4.append(threeAndfour)

    df_run1and2 = pd.concat(run1and2)
    df_run3and4 = pd.concat(run3and4)

    audio_block1and2 = df_run1and2[~np.isnan(df_run1and2.audio)]
    tactile_block1and2 = df_run1and2[~np.isnan(df_run1and2.tactile)]

    fig, ax = plt.subplots(1, 2, sharex='col',figsize=(14,4))

    if 'beep' in folder:
        plt.suptitle('Beeps')
    else:
        plt.suptitle('Graded')

    # Tactile mixed (block 1 and 2) mean RT per probability
    tm_mean_RTs_freqs = tactile_block1and2.groupby(['combinationConditionalProbability']).responseTime.mean()

    tm_meanRT_15 = np.array(tm_mean_RTs_freqs)[0]
    tm_meanRT_35 = np.array(tm_mean_RTs_freqs)[1]
    tm_meanRT_50 = np.array(tm_mean_RTs_freqs)[2]

    # Audio mixed (block 1 and 2) mean RT per probability
    am_mean_RTs_freqs = audio_block1and2.groupby(['combinationConditionalProbability']).responseTime.mean()

    am_meanRT_15 = np.array(am_mean_RTs_freqs)[0]
    am_meanRT_35 = np.array(am_mean_RTs_freqs)[1]
    am_meanRT_50 = np.array(am_mean_RTs_freqs)[2]

    x = ['0.15', '0.35', '0.50']
    tm_meanRT_freqs_list = [tm_meanRT_15, tm_meanRT_35, tm_meanRT_50]

    am_meanRT_freqs_list = [am_meanRT_15, am_meanRT_35, am_meanRT_50]

    ax[0].bar(x, am_meanRT_freqs_list)
    ax[0].set_title("Audio-Visual block 1 and 2 mean RT per probability")
    ax[0].set_ylabel('mean RT')
    ax[0].set_ylim(min(am_meanRT_freqs_list)-0.01, max(am_meanRT_freqs_list)+0.01)


    ax[1].bar(x, tm_meanRT_freqs_list)
    ax[1].set_title("Tactile-Visual block 1 and 2 mean RT per probability")
    ax[1].set_ylabel('mean RT')
    ax[1].set_ylim(min(tm_meanRT_freqs_list)-0.01, max(tm_meanRT_freqs_list)+0.01)

    fig2, ax2 = plt.subplots(1, 2, sharex='col',figsize=(14,4))

    #Audio unique
    audio_block3and4 = df_run3and4[~np.isnan(df_run3and4.audio)]
    au_mean_RTs_freqs = audio_block3and4.groupby(['combinationConditionalProbability']).responseTime.mean()

    #Tactile unique
    tactile_block3and4 = df_run3and4[~np.isnan(df_run3and4.tactile)]
    tu_mean_RTs_freqs = tactile_block3and4.groupby(['combinationConditionalProbability']).responseTime.mean()

    tu_meanRT_15 = np.array(tu_mean_RTs_freqs)[0]
    tu_meanRT_35 = np.array(tu_mean_RTs_freqs)[1]
    tu_meanRT_50 = np.array(tu_mean_RTs_freqs)[2]

    au_meanRT_15 = np.array(au_mean_RTs_freqs)[0]
    au_meanRT_35 = np.array(au_mean_RTs_freqs)[1]
    au_meanRT_50 = np.array(au_mean_RTs_freqs)[2]

    tu_meanRT_freqs_list = [tu_meanRT_15, tu_meanRT_35, tu_meanRT_50]

    au_meanRT_freqs_list = [au_meanRT_15, au_meanRT_35, au_meanRT_50]

    ax2[0].bar(x, au_meanRT_freqs_list)
    ax2[0].set_title("Audio-Visual block 3/4 mean RT per probability")
    ax2[0].set_ylabel('mean RT')
    ax2[0].set_ylim(min(au_meanRT_freqs_list)-0.01, max(au_meanRT_freqs_list)+0.01)

    ax2[1].bar(x, tu_meanRT_freqs_list)
    ax2[1].set_title("Tactile-Visual block 3/4 mean RT per probability")
    ax2[1].set_ylabel('mean RT')
    ax2[1].set_ylim(min(tu_meanRT_freqs_list)-0.01, max(tu_meanRT_freqs_list)+0.01)

    if group:
        fig3, ax3 = plt.subplots(1, 2, sharex='col',figsize=(14,4))
        all_audio = pd.concat((audio_block1and2, audio_block3and4))
        all_tactile = pd.concat((tactile_block1and2, tactile_block3and4))

        at_mean_RTs_freqs = all_audio.groupby(['combinationConditionalProbability']).responseTime.mean()
        at_mean_RTs_stds = all_audio.groupby(['combinationConditionalProbability']).responseTime.std()
        tt_mean_RTs_freqs = all_tactile.groupby(['combinationConditionalProbability']).responseTime.mean()
        tt_mean_RTs_stds = all_tactile.groupby(['combinationConditionalProbability']).responseTime.std()

        at_meanRT_15 = np.array(at_mean_RTs_freqs)[0]
        at_meanRT_35 = np.array(at_mean_RTs_freqs)[1]
        at_meanRT_50 = np.array(at_mean_RTs_freqs)[2]

        tt_meanRT_15 = np.array(tt_mean_RTs_freqs)[0]
        tt_meanRT_35 = np.array(tt_mean_RTs_freqs)[1]
        tt_meanRT_50 = np.array(tt_mean_RTs_freqs)[2]

        at_meanRT_freqs_list = [at_meanRT_15, at_meanRT_35, at_meanRT_50]
        at_error_bars = [np.array(at_mean_RTs_stds)[0], np.array(at_mean_RTs_stds)[1], np.array(at_mean_RTs_stds)[2]]
        tt_meanRT_freqs_list = [tt_meanRT_15, tt_meanRT_35, tt_meanRT_50]
        tt_error_bars = [np.array(tt_mean_RTs_stds)[0], np.array(tt_mean_RTs_stds)[1], np.array(tt_mean_RTs_stds)[2]]

        ax3[0].bar(x, at_meanRT_freqs_list, yerr=at_error_bars)
        ax3[0].set_title("Audio-Visual mean RT per probability")
        ax3[0].set_ylabel('mean RT')
        #ax3[0].set_ylim(min(at_meanRT_freqs_list)-0.01, max(at_meanRT_freqs_list)+0.01)

        ax3[1].bar(x, tt_meanRT_freqs_list, yerr=tt_error_bars)
        ax3[1].set_title("Tactile-Visual mean RT per probability")
        ax3[1].set_ylabel('mean RT')
        #ax3[1].set_ylim(min(tt_meanRT_freqs_list)-0.01, max(tt_meanRT_freqs_list)+0.01)

    plt.show()

```

```python
def rankIDs():

    folder_beeps = 'Data/Pilot/beepStimuli'
    folder_graded = 'Data/Pilot/gradedStimuli'

    dir_file = pathlib.Path().absolute().parent.parent.resolve()
    wanted_dir_beeps = os.path.join(dir_file, folder_beeps)
    wanted_dir_graded = os.path.join(dir_file, folder_graded)

    # Get all savedVals files
    files_beeps = [os.path.join(root, name) for root, dirs, files in os.walk(wanted_dir_beeps) for name in files if name.endswith('savedValues.csv')]
    files_graded = [os.path.join(root, name) for root, dirs, files in os.walk(wanted_dir_graded) for name in files if name.endswith('savedValues.csv')]

    # Beeps - Loop through files and store final accuracy val
    run1and2 = list()
    run3and4 = list()
    count_b = 0
    for file in files_beeps:
        data = pd.read_csv(file)
        parID = re.findall(r'\d+', file)[2]
        one_acc = np.array(data[data.runNumber == 1][['totalAccuracy']])[-1]
        two_acc = np.array(data[data.runNumber == 2][['totalAccuracy']])[-1]
        oneAndtwo = np.mean([one_acc, two_acc])

        three_acc = np.array(data[data.runNumber == 3][['totalAccuracy']])[-1]
        four_acc = np.array(data[data.runNumber == 4][['totalAccuracy']])[-1]
        threeAndfour = np.mean([three_acc, four_acc])

        run1and2.append(pd.DataFrame(data=[[parID,oneAndtwo]], index=[count_b], columns=['ID','meanAcc']))
        run3and4.append(pd.DataFrame(data=[[parID,threeAndfour]], index=[count_b], columns=['ID','meanAcc']))
        count_b += 1

    df_run1and2 = pd.concat(run1and2).sort_values(by='meanAcc')
    df_run3and4 = pd.concat(run3and4).sort_values(by='meanAcc')

    # No beeps - Loop through files and store final accuracy val

    run1and2_g = list()
    run3and4_g = list()
    count_g = 0
    for file in files_graded:
        data_g = pd.read_csv(file)
        parID_g = re.findall(r'\d+', file)[2]
        one_acc_g = np.array(data_g[data_g.runNumber == 1][['totalAccuracy']])[-1]
        two_acc_g = np.array(data_g[data_g.runNumber == 2][['totalAccuracy']])[-1]
        oneAndtwo_g = np.mean([one_acc_g, two_acc_g])

        three_acc_g = np.array(data_g[data_g.runNumber == 3][['totalAccuracy']])[-1]
        four_acc_g = np.array(data_g[data_g.runNumber == 4][['totalAccuracy']])[-1]
        threeAndfour_g = np.mean([three_acc_g, four_acc_g])

        run1and2_g.append(pd.DataFrame(data=[[parID_g,oneAndtwo_g]], index=[count_g], columns=['ID','meanAcc']))
        run3and4_g.append(pd.DataFrame(data=[[parID_g,threeAndfour_g]], index=[count_g], columns=['ID','meanAcc']))
        count_g += 1

    df_run1and2_g = pd.concat(run1and2_g).sort_values(by='meanAcc')
    df_run3and4_g = pd.concat(run3and4_g).sort_values(by='meanAcc')

    # Plot everything

    fig, ax = plt.subplots(2, 2,sharey='col', figsize=(14,8))
    plt.suptitle("Participants ranked by accuracy")

    ax[0,0].bar(x=df_run1and2.ID, height=df_run1and2.meanAcc)
    ax[0,0].axhline(df_run1and2.meanAcc.mean(), color='blue', linewidth=2)
    ax[0,0].set_title('Beeps')
    #ax[0,0].set_xlabel('ParticipantID')
    ax[0,0].set_ylabel('Intermixed accuracy')
    ax[0,0].set_ylim(0.5, 0.95)


    ax[1,0].bar(x=df_run3and4.ID, height=df_run3and4.meanAcc)
    ax[1,0].axhline(df_run3and4.meanAcc.mean(), color='blue', linewidth=2)
    ax[1,0].set_xlabel('ParticipantID')
    ax[1,0].set_ylabel('Pure accuracy')
    ax[1,0].set_ylim(0.5, 0.95)

    ax[0,1].bar(x=df_run1and2_g.ID, height=df_run1and2_g.meanAcc)
    ax[0,1].axhline(df_run1and2_g.meanAcc.mean(), color='blue', linewidth=2)
    ax[0,1].set_title('Graded')
    #ax[0,1].set_xlabel('ParticipantID')
    ax[0,1].set_ylim(0.5, 0.95)

    ax[1,1].bar(x=df_run3and4_g.ID, height=df_run3and4_g.meanAcc)
    ax[1,1].axhline(df_run3and4_g.meanAcc.mean(), color='blue', linewidth=2)
    ax[1,1].set_xlabel('ParticipantID')
    ax[1,1].set_ylim(0.5, 0.95)

    plt.show()
```

```python
# FOR LATER

def plotGroupAcc(folder='Data/Pilot/beepStimuli', ww=10, method='valid'):

    dir_file = pathlib.Path().absolute().parent.parent.resolve()
    wanted_dir = os.path.join(dir_file, folder)

    # Get all expInfo and savedVals files
    files = [os.path.join(root, name) for root, dirs, files in os.walk(wanted_dir) for name in files if name.endswith('savedValues.csv')]

    # Loop through files and store accuracy per trial type
    int_au = list()
    int_ta = list()
    pure_au = list()
    pure_ta = list()

    for file in files:
        data = pd.read_csv(file)
        data_block1 = data[data.runNumber == 1]
        data_block2 = data[data.runNumber == 2]

        if ~np.isnan(data_block1.audio[0]):

            audio_df = pd.concat((data_block1[['accurate', 'reward']][0:20], data_block1[['accurate', 'reward']][40:60], data_block2[['accurate', 'reward']][20:40]), ignore_index=True)
            int_au_ind = pd.DataFrame(ma(audio_df.accurate, ww, method)).fillna(method='ffill')

            tactile_df = pd.concat((data_block1[['accurate', 'reward']][20:40], data_block2[['accurate', 'reward']][0:20], data_block2[['accurate', 'reward']][40:60]), ignore_index=True)
            int_ta_ind = pd.DataFrame(ma(tactile_df.accurate, ww, method)).fillna(method='ffill')

        else:

            audio_df = pd.concat((data_block1[['accurate', 'reward']][20:40], data_block2[['accurate', 'reward']][0:20], data_block2[['accurate', 'reward']][40:60]), ignore_index=True)
            int_au_ind = pd.DataFrame(ma(audio_df.accurate, ww, method)).fillna(method='ffill')

            tactile_df = pd.concat((data_block1[['accurate', 'reward']][0:20], data_block1[['accurate', 'reward']][40:60], data_block2[['accurate', 'reward']][20:40]), ignore_index=True)
            int_ta_ind = pd.DataFrame(ma(tactile_df.accurate, ww, method)).fillna(method='ffill')

        data_block3 = data[data.runNumber == 3].reset_index(drop=True)
        data_block4 = data[data.runNumber == 4].reset_index(drop=True)

        if ~np.isnan(data_block3.audio[59]):
            pure_au_ind = pd.DataFrame(ma(data_block3.accurate,ww, method)).fillna(method='ffill')
            pure_ta_ind = pd.DataFrame(ma(data_block4.accurate,ww, method)).fillna(method='ffill')

        else:
            pure_ta_ind = pd.DataFrame(ma(data_block3.accurate,ww, method)).fillna(method='ffill')
            pure_au_ind = pd.DataFrame(ma(data_block4.accurate,ww, method)).fillna(method='ffill')

        int_au.append(int_au_ind)
        int_ta.append(int_ta_ind)
        pure_au.append(pure_au_ind)
        pure_ta.append(pure_ta_ind)

    df_int_au = pd.concat(int_au, axis=1)
    df_int_ta = pd.concat(int_ta, axis=1)
    df_pure_au = pd.concat(pure_au, axis=1)
    df_pure_ta = pd.concat(pure_ta, axis=1)

    df_int_au_avg = np.mean(df_int_au, axis=1)
    df_int_ta_avg = np.mean(df_int_ta, axis=1)
    df_pure_au_avg = np.mean(df_pure_au, axis=1)
    df_pure_ta_avg = np.mean(df_pure_ta, axis=1)


    fig, ax = plt.subplots(2, 2,sharey='col', figsize=(14,8))
    plt.suptitle("Average learning curves")
    ax[0,0].plot(df_int_au_avg)
    ax[0,0].set_title("Audio-Visual accuracy (intermixed, moving avg: {})".format(ww))
    ax[0,0].set_ylabel('accuracy')
    ax[0,0].set_ylim(0, 1.1)

    ax[0,1].plot(df_int_ta_avg)
    ax[0,1].set_title("Visual-Tactile accuracy (intermixed, moving avg: {})".format(ww))
    ax[0,1].set_ylabel('accuracy')
    ax[0,1].set_ylim(0, 1.1)

    ax[1,0].plot(df_pure_au_avg)
    ax[1,0].set_title("Audio-Visual accuracy (Long, moving avg: {})".format(ww))
    ax[1,0].set_ylabel('accuracy')
    ax[1,0].set_ylim(0, 1.1)

    ax[1,1].plot(df_pure_ta_avg)
    ax[1,1].set_title("Visual-Tactile accuracy (Long, moving avg: {})".format(ww))
    ax[1,1].set_ylabel('accuracy')
    ax[1,1].set_ylim(0, 1.1)

    plt.show()
```

```python
def corr_audio_tactile(folder='Data/Pilot/beepStimuli'):
    dir_file = pathlib.Path().absolute().parent.parent.resolve()
    wanted_dir = os.path.join(dir_file, folder)

    # Get all expInfo and savedVals files
    files = [os.path.join(root, name) for root, dirs, files in os.walk(wanted_dir) for name in files if name.endswith('savedValues.csv')]

    acc_values = list()
    count = 0
    for file in files:
        data = pd.read_csv(file)
        parID = re.findall(r'\d+', file)[2]

        if ~np.isnan(data[data.runNumber == 3].audio[130]):
            audio_acc = np.array(data[data.runNumber == 3][['totalAccuracy']])[-1]
            tactile_acc = np.array(data[data.runNumber == 4][['totalAccuracy']])[-1]

        else:
            audio_acc = np.array(data[data.runNumber == 4][['totalAccuracy']])[-1]
            tactile_acc = np.array(data[data.runNumber == 3][['totalAccuracy']])[-1]

        acc_values.append(pd.DataFrame(data=[[parID,float(audio_acc), float(tactile_acc)]], index=[count], columns=['ID','AudioAcc', 'TactileAcc']))
        count += 1

    df_acc_values = pd.concat(acc_values)
    r, p = stats.pearsonr(df_acc_values['AudioAcc'], df_acc_values['TactileAcc'])
    sns.lmplot(x="AudioAcc", y="TactileAcc", data=df_acc_values)

    return "Pearson's R: {0}, p-value: {1}".format(r, p)
```

```python
# For individual mean RTs

def plotRT(data, info):

    #data_block1 = data[data.runNumber == 1]
    #data_block2 = data[data.runNumber == 2]
    data_block1and2 = data[(data.runNumber == 1) | (data.runNumber == 2)]
    data_block3 = data[data.runNumber == 3]
    data_block4 = data[data.runNumber == 4]

    #df_block1 = pd.DataFrame(data_block1)
    #df_block2 = pd.DataFrame(data_block2)
    df_block1and2 = pd.DataFrame(data_block1and2)
    df_block3 = pd.DataFrame(data_block3)
    df_block4 = pd.DataFrame(data_block4)

    participantID = info.participantID[0]
    Freq0A = info.modality0AFrequency[0]
    Freq1A = info.modality1AFrequency[0]
    Freq2A = info.modality2AFrequency[0]
    Freq0B = info.modality0BFrequency[0]
    Freq1B = info.modality1BFrequency[0]
    Freq2B = info.modality2BFrequency[0]
    Freq0C = info.modality0CFrequency[0]
    Freq1C = info.modality1CFrequency[0]
    Freq2C = info.modality2CFrequency[0]


    fig, ax = plt.subplots(1, 2, sharex='col', sharey='row', figsize=(14,4))

    if info.beeps[0] == 1:
        plt.suptitle('ParticipantID: {}, Beeps'.format(participantID))
    else:
        plt.suptitle('ParticipantID: {}, No beeps'.format(participantID))

    # Tactile mixed (block 1 and 2)
    tm_mean_RTs = df_block1and2.groupby(['tactile', 'visual']).responseTime.mean()
    tm_meanRT_0A = tm_mean_RTs[0][0]
    tm_meanRT_1A = tm_mean_RTs[1][0]
    tm_meanRT_2A = tm_mean_RTs[2][0]

    tm_meanRT_0B = tm_mean_RTs[0][1]
    tm_meanRT_1B = tm_mean_RTs[1][1]
    tm_meanRT_2B = tm_mean_RTs[2][1]

    tm_meanRT_0C = tm_mean_RTs[0][2]
    tm_meanRT_1C = tm_mean_RTs[1][2]
    tm_meanRT_2C = tm_mean_RTs[2][2]

    # Audio mixed (block 1 and 2)
    am_mean_RTs = df_block1and2.groupby(['audio', 'visual']).responseTime.mean()
    am_meanRT_0A = am_mean_RTs[0][0]
    am_meanRT_1A = am_mean_RTs[1][0]
    am_meanRT_2A = am_mean_RTs[2][0]

    am_meanRT_0B = am_mean_RTs[0][1]
    am_meanRT_1B = am_mean_RTs[1][1]
    am_meanRT_2B = am_mean_RTs[2][1]

    am_meanRT_0C = am_mean_RTs[0][2]
    am_meanRT_1C = am_mean_RTs[1][2]
    am_meanRT_2C = am_mean_RTs[2][2]

    x = ['0A: {}'.format(str(Freq0A)[1:]), '1A: {}'.format(str(Freq1A)[1:]), '2A: {}'.format(str(Freq2A)[1:]), '0B: {}'.format(str(Freq0B)[1:]), '1B: {}'.format(str(Freq1B)[1:]), '2B: {}'.format(str(Freq2B)[1:]), '0C: {}'.format(str(Freq0C)[1:]), '1C: {}'.format(str(Freq1C)[1:]), '2C: {}'.format(str(Freq2C)[1:])]
    tm_meanRT_list = [tm_meanRT_0A, tm_meanRT_1A, tm_meanRT_2A, tm_meanRT_0B, tm_meanRT_1B, tm_meanRT_2B, tm_meanRT_0C, tm_meanRT_1C, tm_meanRT_2C]

    am_meanRT_list = [am_meanRT_0A, am_meanRT_1A, am_meanRT_2A, am_meanRT_0B, am_meanRT_1B, am_meanRT_2B, am_meanRT_0C, am_meanRT_1C, am_meanRT_2C]

    ax[0].bar(x, am_meanRT_list)
    ax[0].set_title("Audio-Visual block 1 and 2 mean RT per combination")
    ax[0].set_ylabel('mean RT')
    ax[0].set_ylim(0, 3)

    ax[1].bar(x, tm_meanRT_list)
    ax[1].set_title("Tactile-Visual block 1 and 2 mean RT per combination")
    ax[1].set_ylabel('mean RT')
    ax[1].set_ylim(0, 3)



    fig2, ax2 = plt.subplots(1, 2, sharex='col', sharey='row',figsize=(14,4))

    # Tactile unique (block 3 or 4)
    if ~np.isnan(data_block3.tactile[120]):
        tu_mean_RTs = df_block3.groupby(['tactile', 'visual']).responseTime.mean()
    else:
        tu_mean_RTs = df_block4.groupby(['tactile', 'visual']).responseTime.mean()
    tu_meanRT_0A = tu_mean_RTs[0][0]
    tu_meanRT_1A = tu_mean_RTs[1][0]
    tu_meanRT_2A = tu_mean_RTs[2][0]

    tu_meanRT_0B = tu_mean_RTs[0][1]
    tu_meanRT_1B = tu_mean_RTs[1][1]
    tu_meanRT_2B = tu_mean_RTs[2][1]

    tu_meanRT_0C = tu_mean_RTs[0][2]
    tu_meanRT_1C = tu_mean_RTs[1][2]
    tu_meanRT_2C = tu_mean_RTs[2][2]

    # Audio unique (block 3 or 4)
    if np.isnan(data_block3.tactile[120]):
        au_mean_RTs = df_block3.groupby(['audio', 'visual']).responseTime.mean()
    else:
        au_mean_RTs = df_block4.groupby(['audio', 'visual']).responseTime.mean()

    au_meanRT_0A = au_mean_RTs[0][0]
    au_meanRT_1A = au_mean_RTs[1][0]
    au_meanRT_2A = au_mean_RTs[2][0]

    au_meanRT_0B = au_mean_RTs[0][1]
    au_meanRT_1B = au_mean_RTs[1][1]
    au_meanRT_2B = au_mean_RTs[2][1]

    au_meanRT_0C = au_mean_RTs[0][2]
    au_meanRT_1C = au_mean_RTs[1][2]
    au_meanRT_2C = au_mean_RTs[2][2]

    tu_meanRT_list = [tu_meanRT_0A, tu_meanRT_1A, tu_meanRT_2A, tu_meanRT_0B, tu_meanRT_1B, tu_meanRT_2B, tu_meanRT_0C, tu_meanRT_1C, tu_meanRT_2C]

    au_meanRT_list = [au_meanRT_0A, au_meanRT_1A, au_meanRT_2A, au_meanRT_0B, au_meanRT_1B, au_meanRT_2B, au_meanRT_0C, au_meanRT_1C, au_meanRT_2C]

    ax2[0].bar(x, au_meanRT_list)
    ax2[0].set_title("Audio-Visual block 3/4 mean RT per combination")
    ax2[0].set_ylabel('mean RT')
    ax2[0].set_ylim(0, 3)

    ax2[1].bar(x, tu_meanRT_list)
    ax2[1].set_title("Tactile-Visual block 3/4 mean RT per combination")
    ax2[1].set_ylabel('mean RT')
    ax2[1].set_ylim(0, 3)

    fig3, ax3 = plt.subplots(1, 2, sharex='col', sharey='row',figsize=(14,4))

    audio_block1and2 = df_block1and2[~np.isnan(df_block1and2.audio)]
    tactile_block1and2 = df_block1and2[~np.isnan(df_block1and2.tactile)]

    # Tactile mixed (block 1 and 2) mean RT per probability
    tm_mean_RTs_freqs = tactile_block1and2.groupby(['combinationConditionalProbability']).responseTime.mean()

    tm_meanRT_15 = np.array(tm_mean_RTs_freqs)[0]
    tm_meanRT_35 = np.array(tm_mean_RTs_freqs)[1]
    tm_meanRT_50 = np.array(tm_mean_RTs_freqs)[2]

    # Audio mixed (block 1 and 2) mean RT per probability
    am_mean_RTs_freqs = audio_block1and2.groupby(['combinationConditionalProbability']).responseTime.mean()

    am_meanRT_15 = np.array(am_mean_RTs_freqs)[0]
    am_meanRT_35 = np.array(am_mean_RTs_freqs)[1]
    am_meanRT_50 = np.array(am_mean_RTs_freqs)[2]

    x = ['0.15', '0.35', '0.50']
    tm_meanRT_freqs_list = [tm_meanRT_15, tm_meanRT_35, tm_meanRT_50]

    am_meanRT_freqs_list = [am_meanRT_15, am_meanRT_35, am_meanRT_50]

    ax3[0].bar(x, tm_meanRT_freqs_list)
    ax3[0].set_title("Audio-Visual block 1 and 2 mean RT per probability")
    ax3[0].set_ylabel('mean RT')
    ax3[0].set_ylim(0, 3)

    ax3[1].bar(x, am_meanRT_freqs_list)
    ax3[1].set_title("Tactile-Visual block 1 and 2 mean RT per probability")
    ax3[1].set_ylabel('mean RT')
    ax3[1].set_ylim(0, 3)

    fig4, ax4 = plt.subplots(1, 2, sharex='col', sharey='row',figsize=(14,4))

    #Audio unique
    if ~np.isnan(data_block3.tactile[120]):
        tu_mean_RTs_freqs = df_block3.groupby(['combinationConditionalProbability']).responseTime.mean()
    else:
        tu_mean_RTs_freqs = df_block4.groupby(['combinationConditionalProbability']).responseTime.mean()

    #Tactile unique
    if np.isnan(data_block3.tactile[120]):
        au_mean_RTs_freqs = df_block3.groupby(['combinationConditionalProbability']).responseTime.mean()
    else:
        au_mean_RTs_freqs = df_block4.groupby(['combinationConditionalProbability']).responseTime.mean()

    tu_meanRT_15 = np.array(tu_mean_RTs_freqs)[0]
    tu_meanRT_35 = np.array(tu_mean_RTs_freqs)[1]
    tu_meanRT_50 = np.array(tu_mean_RTs_freqs)[2]

    au_meanRT_15 = np.array(au_mean_RTs_freqs)[0]
    au_meanRT_35 = np.array(au_mean_RTs_freqs)[1]
    au_meanRT_50 = np.array(au_mean_RTs_freqs)[2]

    tu_meanRT_freqs_list = [tu_meanRT_15, tu_meanRT_35, tu_meanRT_50]

    au_meanRT_freqs_list = [au_meanRT_15, au_meanRT_35, au_meanRT_50]

    ax4[0].bar(x, tu_meanRT_freqs_list)
    ax4[0].set_title("Audio-Visual block 3/4 mean RT per probability")
    ax4[0].set_ylabel('mean RT')
    ax4[0].set_ylim(0, 3)

    ax4[1].bar(x, au_meanRT_freqs_list)
    ax4[1].set_title("Tactile-Visual block 3/4 mean RT per probability")
    ax4[1].set_ylabel('mean RT')
    ax4[1].set_ylim(0, 3)


    plt.show()
```
