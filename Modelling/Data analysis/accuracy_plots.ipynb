##### Individual learning curves with moving average of 5

```python
def ma(interval, window_size, method):
    window = np.ones(int(window_size))/float(window_size)
    return np.convolve(interval, window, method)
```

```python
def plotAccuracy(data, info, ww, method):

    participantID = info.participantID[0]

    if max(data.runNumber) == 4:
        data_block1 = data[data.runNumber == 1].reset_index(drop=True)
        data_block2 = data[data.runNumber == 2].reset_index(drop=True)
        data_block3 = data[data.runNumber == 3].reset_index(drop=True)
        data_block4 = data[data.runNumber == 4].reset_index(drop=True)

        fig, ax = plt.subplots(2, 2, sharex='col', sharey='row', figsize=(10,4))

        if info.beeps[0] == 1:
            plt.suptitle('ParticipantID: {}, Beeps'.format(participantID))
        else:
            plt.suptitle('ParticipantID: {}, No beeps'.format(participantID))

        if ~np.isnan(data_block1.audio[59]):
            ax[0,0].plot(pd.DataFrame(ma(data_block1.accurate,ww, method)).fillna(method='ffill'))
            ax[0,0].set_title("Audio-Visual accuracy (block 1, moving avg: {})".format(ww))
            ax[0,0].set_ylabel('accuracy')
            ax[0,0].set_ylim(0, 1.1)

            ax[0,1].plot(pd.DataFrame(ma(data_block2.accurate,ww, method)).fillna(method='ffill'))
            ax[0,1].set_title("Tactile-Visual accuracy (block 2, moving avg: {})".format(ww))
            ax[0,1].set_ylabel('accuracy')
            ax[0,1].set_ylim(0, 1.1)

            ax[1,0].plot(pd.DataFrame(ma(data_block3.accurate,ww, method)).fillna(method='ffill'))
            ax[1,0].set_title("Audio-Visual accuracy (block 3, moving avg: {})".format(ww))
            ax[1,0].set_ylabel('accuracy')
            ax[1,0].set_ylim(0, 1.1)

            ax[1,1].plot(pd.DataFrame(ma(data_block4.accurate,ww, method)).fillna(method='ffill'))
            ax[1,1].set_title("Tactile-Visual accuracy (block 4, moving avg: {})".format(ww))
            ax[1,1].set_ylabel('accuracy')
            ax[1,1].set_ylim(0, 1.1)

            if method == 'same':
                audio_linesb1 = data_block1[(data_block1.accurate != data_block1.reward) & (~np.isnan(data_block1.reward))].index.tolist()
                tactile_linesb2 = data_block2[(data_block2.accurate != data_block2.reward) & (~np.isnan(data_block2.reward))].index.tolist()
                audio_linesb3 = data_block3[(data_block3.accurate != data_block3.reward) & (~np.isnan(data_block3.reward))].index.tolist()
                tactile_linesb4 = data_block4[(data_block4.accurate != data_block4.reward) & (~np.isnan(data_block4.reward))].index.tolist()

                for xc in audio_linesb1:
                    ax[0,0].axvline(x=xc, color='k', linestyle='--', linewidth=1)
                for xc in tactile_linesb2:
                    ax[0,1].axvline(x=xc, color='k', linestyle='--', linewidth=1)
                for xc in audio_linesb3:
                    ax[1,0].axvline(x=xc, color='k', linestyle='--', linewidth=1)
                for xc in tactile_linesb4:
                    ax[1,1].axvline(x=xc, color='k', linestyle='--', linewidth=1)

        else:
            ax[0,0].plot(pd.DataFrame(ma(data_block1.accurate,ww, method)).fillna(method='ffill'))
            ax[0,0].set_title("Tactile-Visual accuracy (block 1, moving avg: {})".format(ww))
            ax[0,0].set_ylabel('accuracy')
            ax[0,0].set_ylim(0, 1.1)

            ax[0,1].plot(pd.DataFrame(ma(data_block2.accurate,ww, method)).fillna(method='ffill'))
            ax[0,1].set_title("Audio-Visual accuracy (block 2, moving avg: {})".format(ww))
            ax[0,1].set_ylabel('accuracy')
            ax[0,1].set_ylim(0, 1.1)

            ax[1,0].plot(pd.DataFrame(ma(data_block3.accurate,ww, method)).fillna(method='ffill'))
            ax[1,0].set_title("Tactile-Visual accuracy (block 3, moving avg: {})".format(ww))
            ax[1,0].set_ylabel('accuracy')
            ax[1,0].set_ylim(0, 1.1)

            ax[1,1].plot(pd.DataFrame(ma(data_block4.accurate,ww, method)).fillna(method='ffill'))
            ax[1,1].set_title("Audio-Visual accuracy (block 4, moving avg: {})".format(ww))
            ax[1,1].set_ylabel('accuracy')
            ax[1,1].set_ylim(0, 1.1)

            if method == 'same':
                tactile_linesb1 = data_block1[(data_block1.accurate != data_block1.reward) & (~np.isnan(data_block1.reward))].index.tolist()
                audio_linesb2 = data_block2[(data_block2.accurate != data_block2.reward) & (~np.isnan(data_block2.reward))].index.tolist()
                tactile_linesb3 = data_block3[(data_block3.accurate != data_block3.reward) & (~np.isnan(data_block3.reward))].index.tolist()
                audio_linesb4 = data_block4[(data_block4.accurate != data_block4.reward) & (~np.isnan(data_block4.reward))].index.tolist()

                for xc in audio_linesb2:
                    ax[0,1].axvline(x=xc, color='k', linestyle='--', linewidth=1)
                for xc in tactile_linesb1:
                    ax[0,0].axvline(x=xc, color='k', linestyle='--', linewidth=1)
                for xc in audio_linesb4:
                    ax[1,1].axvline(x=xc, color='k', linestyle='--', linewidth=1)
                for xc in tactile_linesb3:
                    ax[1,0].axvline(x=xc, color='k', linestyle='--', linewidth=1)

    else:
        data_block1 = data[data.runNumber == 1].reset_index(drop=True)
        data_block2 = data[data.runNumber == 2].reset_index(drop=True)

        fig, ax = plt.subplots(1, 2, sharex='col', sharey='row', figsize=(10,4))

        if info.beeps[0] == 1:
            plt.suptitle('ParticipantID: {}, Beeps'.format(participantID))
        else:
            plt.suptitle('ParticipantID: {}, No beeps'.format(participantID))


        if ~np.isnan(data_block1.audio[59]):
            ax[0].plot(pd.DataFrame(ma(data_block1.accurate,ww, method)).fillna(method='ffill'))
            ax[0].set_title("Audio-Visual accuracy (block 1, moving avg: {})".format(ww))
            ax[0].set_ylabel('accuracy')
            ax[0].set_ylim(0, 1.1)

            ax[1].plot(pd.DataFrame(ma(data_block2.accurate,ww, method)).fillna(method='ffill'))
            ax[1].set_title("Tactile-Visual accuracy (block 2, moving avg: {})".format(ww))
            ax[1].set_ylabel('accuracy')
            ax[1].set_ylim(0, 1.1)

            if method == 'same':
                audio_linesb1 = data_block1[(data_block1.accurate != data_block1.reward) & (~np.isnan(data_block1.reward))].index.tolist()
                tactile_linesb2 = data_block2[(data_block2.accurate != data_block2.reward) & (~np.isnan(data_block2.reward))].index.tolist()

                for xc in audio_linesb1:
                    ax[0].axvline(x=xc, color='k', linestyle='--', linewidth=1)
                for xc in tactile_linesb2:
                    ax[1].axvline(x=xc, color='k', linestyle='--', linewidth=1)

        else:

            ax[0].plot(pd.DataFrame(ma(data_block1.accurate,ww, method)).fillna(method='ffill'))
            ax[0].set_title("Tactile-Visual accuracy (block 1, moving avg: {})".format(ww))
            ax[0].set_ylabel('accuracy')
            ax[0].set_ylim(0, 1.1)

            ax[1].plot(pd.DataFrame(ma(data_block2.accurate,ww, method)).fillna(method='ffill'))
            ax[1].set_title("Audio-Visual accuracy (block 2, moving avg: {})".format(ww))
            ax[1].set_ylabel('accuracy')
            ax[1].set_ylim(0, 1.1)

            if method == 'same':
                tactile_linesb1 = data_block1[(data_block1.accurate != data_block1.reward) & (~np.isnan(data_block1.reward))].index.tolist()
                audio_linesb2 = data_block2[(data_block2.accurate != data_block2.reward) & (~np.isnan(data_block2.reward))].index.tolist()

                for xc in audio_linesb2:
                    ax[1].axvline(x=xc, color='k', linestyle='--', linewidth=1)
                for xc in tactile_linesb1:
                    ax[0].axvline(x=xc, color='k', linestyle='--', linewidth=1)

    plt.show()

```

##### Mean RT group analysis

```python
def plotGroupRT(folder='Data/Pilot/beepStimuli/modified_files', exclude=False):

    dir_file = pathlib.Path().absolute().parent.parent.resolve()
    wanted_dir = os.path.join(dir_file, folder)

    # Get all expInfo and savedVals files
    files = [os.path.join(root, name) for root, dirs, files in os.walk(wanted_dir) for name in files if name.endswith('savedValues.csv')]

    # Loop through files and store RT vals, audio, tactile with probability per run and trial type
    audioRuns = list()
    tactileRuns = list()

    for file in files:
        data = pd.read_csv(file)
        if max(data.runNumber) == 4:
            if ~np.isnan(data.audio[59]):
                audios = data[(data.runNumber == 1) | (data.runNumber == 3)][['responseTime', 'audio', 'tactile', 'combinationConditionalProbability']]
                tactiles = data[(data.runNumber == 2) | (data.runNumber == 4)][['responseTime', 'audio', 'tactile', 'combinationConditionalProbability']]
            else:
                tactiles = data[(data.runNumber == 1) | (data.runNumber == 3)][['responseTime', 'audio', 'tactile', 'combinationConditionalProbability']]
                audios = data[(data.runNumber == 2) | (data.runNumber == 4)][['responseTime', 'audio', 'tactile', 'combinationConditionalProbability']]
        else:
            if ~np.isnan(data.audio[59]):
                audios = data[data.runNumber == 1][['responseTime', 'audio', 'tactile', 'combinationConditionalProbability']]
                tactiles = data[data.runNumber == 2][['responseTime', 'audio', 'tactile', 'combinationConditionalProbability']]
            else:
                tactiles = data[data.runNumber == 1][['responseTime', 'audio', 'tactile', 'combinationConditionalProbability']]
                audios = data[data.runNumber == 2][['responseTime', 'audio', 'tactile', 'combinationConditionalProbability']]

        if exclude:
            if max(data.runNumber) == 4:
                one_acc = np.array(data[data.runNumber == 1][['totalAccuracy']])[-1]
                two_acc = np.array(data[data.runNumber == 2][['totalAccuracy']])[-1]
                three_acc = np.array(data[data.runNumber == 3][['totalAccuracy']])[-1]
                four_acc = np.array(data[data.runNumber == 4][['totalAccuracy']])[-1]
                avg_acc = np.mean((one_acc, two_acc,three_acc, four_acc))
            else:
                one_acc = np.array(data[data.runNumber == 1][['totalAccuracy']])[-1]
                two_acc = np.array(data[data.runNumber == 2][['totalAccuracy']])[-1]
                avg_acc = np.mean((one_acc, two_acc))
            if avg_acc >= 0.6:
                audioRuns.append(audios)
                tactileRuns.append(tactiles)
        else:
            tactileRuns.append(tactiles)
            audioRuns.append(audios)

    df_audioRuns = pd.concat(audioRuns)
    df_tactileRuns = pd.concat(tactileRuns)

    fig, ax = plt.subplots(1, 2, sharex='col',figsize=(14,4))

    #Audio
    au_mean_RTs_freqs = df_audioRuns.groupby(['combinationConditionalProbability']).responseTime.mean()
    au_mean_RTs_stds = df_audioRuns.groupby(['combinationConditionalProbability']).responseTime.std()

    #Tactile
    ta_mean_RTs_freqs = df_tactileRuns.groupby(['combinationConditionalProbability']).responseTime.mean()

    ta_mean_RTs_stds = df_tactileRuns.groupby(['combinationConditionalProbability']).responseTime.std()

    ta_meanRT_15 = np.array(ta_mean_RTs_freqs)[0]
    ta_meanRT_35 = np.array(ta_mean_RTs_freqs)[1]
    ta_meanRT_50 = np.array(ta_mean_RTs_freqs)[2]

    au_meanRT_15 = np.array(au_mean_RTs_freqs)[0]
    au_meanRT_35 = np.array(au_mean_RTs_freqs)[1]
    au_meanRT_50 = np.array(au_mean_RTs_freqs)[2]

    ta_meanRT_freqs_list = [ta_meanRT_15, ta_meanRT_35, ta_meanRT_50]
    ta_error_bars = [np.array(ta_mean_RTs_stds)[0], np.array(ta_mean_RTs_stds)[1], np.array(ta_mean_RTs_stds)[2]]

    au_meanRT_freqs_list = [au_meanRT_15, au_meanRT_35, au_meanRT_50]
    au_error_bars = [np.array(au_mean_RTs_stds)[0], np.array(au_mean_RTs_stds)[1], np.array(au_mean_RTs_stds)[2]]

    x = ['0.15', '0.35', '0.50']

    ax[0].bar(x, au_meanRT_freqs_list, yerr=au_error_bars)
    ax[0].set_title("Audio-Visual mean RT per probability")
    ax[0].set_ylabel('mean RT')
    ax[0].set_ylim(1, 2)

    ax[1].bar(x, ta_meanRT_freqs_list, yerr=ta_error_bars)
    ax[1].set_title("Tactile-Visual mean RT per probability")
    ax[1].set_ylabel('mean RT')
    ax[1].set_ylim(1, 2)


    plt.show()

```

##### Rank participants based on accuracy

```python
def rankIDs():

    folder_beeps = 'Data/Pilot/beepStimuli/modified_files'

    dir_file = pathlib.Path().absolute().parent.parent.resolve()
    wanted_dir_beeps = os.path.join(dir_file, folder_beeps)

    # Get all savedVals files
    files_beeps = [os.path.join(root, name) for root, dirs, files in os.walk(wanted_dir_beeps) for name in files if name.endswith('savedValues.csv')]

    # Beeps - Loop through files and store final accuracy val
    audioRuns = list()
    tactileRuns = list()
    count_b = 0
    for file in files_beeps:
        data = pd.read_csv(file)
        parID = re.findall(r'\d+', file)[0]
        if ~np.isnan(data.audio[59]):
            if max(data.runNumber == 4):
                one_audio = np.array(data[data.runNumber == 1][['totalAccuracy']])[-1]
                two_audio = np.array(data[data.runNumber == 3][['totalAccuracy']])[-1]
                all_audio = np.mean([one_audio, two_audio])
                one_tactile = np.array(data[data.runNumber == 2][['totalAccuracy']])[-1]
                two_tactile = np.array(data[data.runNumber == 4][['totalAccuracy']])[-1]
                all_tactile = np.mean([one_tactile, two_tactile])
            else:
                all_audio = np.array(data[data.runNumber == 1][['totalAccuracy']])[-1]
                all_tactile = np.array(data[data.runNumber == 2][['totalAccuracy']])[-1]
        else:
            if max(data.runNumber == 4):
                one_audio = np.array(data[data.runNumber == 2][['totalAccuracy']])[-1]
                two_audio = np.array(data[data.runNumber == 4][['totalAccuracy']])[-1]
                all_audio = np.mean([one_audio, two_audio])
                one_tactile = np.array(data[data.runNumber == 1][['totalAccuracy']])[-1]
                two_tactile = np.array(data[data.runNumber == 3][['totalAccuracy']])[-1]
                all_tactile = np.mean([one_tactile, two_tactile])
            else:
                all_audio = np.array(data[data.runNumber == 2][['totalAccuracy']])[-1]
                all_tactile = np.array(data[data.runNumber == 1][['totalAccuracy']])[-1]


        audioRuns.append(pd.DataFrame(data=[[parID,all_audio]], index=[count_b], columns=['ID','meanAcc']))
        tactileRuns.append(pd.DataFrame(data=[[parID,all_tactile]], index=[count_b], columns=['ID','meanAcc']))
        count_b += 1

    df_audioRuns = pd.concat(audioRuns).sort_values(by='meanAcc')
    df_tactileRuns = pd.concat(tactileRuns).sort_values(by='meanAcc')

    # Plot everything

    fig, ax = plt.subplots(2, 1,sharey='col', figsize=(14,8))
    plt.suptitle("Participants ranked by accuracy")

    ax[0].bar(x=df_audioRuns.ID, height=df_audioRuns.meanAcc)
    ax[0].axhline(df_audioRuns.meanAcc.mean(), color='blue', linewidth=2)
    ax[0].set_xlabel('ParticipantID')
    ax[0].set_ylabel('Audio-visual accuracy')
    ax[0].set_ylim(0.5, 0.95)


    ax[1].bar(x=df_tactileRuns.ID, height=df_tactileRuns.meanAcc)
    ax[1].axhline(df_tactileRuns.meanAcc.mean(), color='blue', linewidth=2)
    ax[1].set_xlabel('ParticipantID')
    ax[1].set_ylabel('Tactile-visual accuracy')
    ax[1].set_ylim(0.5, 0.95)

    plt.show()
```

```python
def corr_audio_tactile(folder='Data/Pilot/beepStimuli/modified_files'):
    dir_file = pathlib.Path().absolute().parent.parent.resolve()
    wanted_dir = os.path.join(dir_file, folder)

    # Get all expInfo and savedVals files
    files = [os.path.join(root, name) for root, dirs, files in os.walk(wanted_dir) for name in files if name.endswith('savedValues.csv')]

    acc_values = list()
    count = 0
    for file in files:
        data = pd.read_csv(file)
        parID = re.findall(r'\d+', file)[0]

        if max(data.runNumber) == 4:
            if ~np.isnan(data.audio[59]):
                audio_acc = np.array(data[(data.runNumber == 1) | (data.runNumber == 3)][['totalAccuracy']])[-1]
                tactile_acc = np.array(data[(data.runNumber == 2) | (data.runNumber == 4)][['totalAccuracy']])[-1]
            else:
                audio_acc = np.array(data[(data.runNumber == 2) | (data.runNumber == 4)][['totalAccuracy']])[-1]
                tactile_acc = np.array(data[(data.runNumber == 1) | (data.runNumber == 3)][['totalAccuracy']])[-1]
        else:
            if ~np.isnan(data.audio[59]):
                audio_acc = np.array(data[data.runNumber == 1][['totalAccuracy']])[-1]
                tactile_acc = np.array(data[data.runNumber == 2][['totalAccuracy']])[-1]
            else:
                audio_acc = np.array(data[data.runNumber == 2][['totalAccuracy']])[-1]
                tactile_acc = np.array(data[data.runNumber == 1][['totalAccuracy']])[-1]


        acc_values.append(pd.DataFrame(data=[[parID,float(audio_acc), float(tactile_acc)]], index=[count], columns=['ID','AudioAcc', 'TactileAcc']))
        count += 1

    df_acc_values = pd.concat(acc_values)
    r, p = stats.pearsonr(df_acc_values['AudioAcc'], df_acc_values['TactileAcc'])
    sns.lmplot(x="AudioAcc", y="TactileAcc", data=df_acc_values)

    return "Pearson's R: {0}, p-value: {1}".format(r, p)
```
