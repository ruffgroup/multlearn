```python
from nilearn import image, plotting
```

```python
ag_s_l = image.load_img('/data/ds-mlearn/nipype/model2/ROI/AG_S_L.nii')
ag_rpe_l = image.load_img('/data/ds-mlearn/nipype/model2/ROI/AG_RPE_L.nii')
```

```python
overlap = image.math_img('img1 & img2', img1=ag_s_l, img2=ag_rpe_l)
overlap.to_filename('/data/ds-mlearn/nipype/model2/ROI/AG_S_L_RPE_L_overlap.nii')
```

```python
plotting.plot_roi(overlap, title='Overlap of AG_S_L and AG_RPE_L', )
```
